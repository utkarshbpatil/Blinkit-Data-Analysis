```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
```

```python
df['item_fat_content'] = df['item_fat_content'].replace({
    'LF': 'Low Fat',
    'low fat': 'Low Fat',
    'reg': 'Regular'
})
```

```python
df['item_weight'] = df['item_weight'].fillna(df['item_weight'].median())
```
